```python
from torch import nn
from torchvision import models
import ml_modules
```

# ViT B 16

```python
weights = models.ViT_B_16_Weights.DEFAULT
model = models.vit_b_16(weights=weights)
```

```python
model
```

```python
model.encoder.ln
```

```python
## Freeze parameters

for params in model.parameters():

    params.requires_grad = False
```

```python
## Classifier head

model.heads = nn.Sequential(
    nn.Linear(in_features=768,
              out_features=6,
              bias=True)
)
```

```python
## Unfreeze encoder_layer_11

for params in model.encoder.layers.encoder_layer_11.parameters():

    params.requires_grad = True

for params in model.encoder.ln.parameters():

    params.requires_grad = True
```

```python
ml_modules.model_summary(model=model,
                         input_size=[32, 3, 224, 224])
```
